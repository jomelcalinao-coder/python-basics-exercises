### 2. Check if the first and last numbers of a list are the same

Write a code to return True if the list’s first and last numbers are the same. If the numbers are different, return False.

Given:

numbers_x = [10, 20, 30, 40, 10]
Output True

numbers_y = [75, 65, 35, 75, 30]
Output False



```python
def check_first_last_same(numbers):
    return numbers[0] == numbers[-1]


numbers_x = [10, 20, 30, 40, 10]
numbers_y = [75, 65, 35, 75, 30]

print(check_first_last_same(numbers_x))  # True
print(check_first_last_same(numbers_y))  # False
```

```python

def first_last_num_equal(numbers):
  if numbers[0] == numbers[-1]:
    return True
  else:
    return False

numbers_x = [10, 20, 30, 40, 10]
numbers_y = [75, 65, 35, 75, 30]

print(first_last_num_equal(numbers_x))
print(first_last_num_equal(numbers_y))
```

3. Merge two lists using the following condition

Given two lists of numbers, write Python code to create a new list containing odd numbers from the first list and even numbers from the second list.

Given:

list1 = [10, 20, 25, 30, 35]
list2 = [40, 45, 60, 75, 90]
Expected Output:

result list: [25, 35, 40, 60, 90]

```python
list1 = [10, 20, 23, 11, 17]
list2 = [13, 43, 24, 36, 12]

result = []

for num in list1:
    if num % 2 != 0:
        result.append(num)

for num in list2:
    if num % 2 == 0:
        result.append(num)

print(result)
```

```python
list1 = [10, 20, 25, 30, 35]
list2 = [40, 45, 60, 75, 90]

def combine_odd_even(list1, list2):
    result = []

    # Add odd numbers from the first list
    for num in list1:
        if num % 2 != 0:
            result.append(num)

    # Add even numbers from the second list
    for num in list2:
        if num % 2 == 0:
            result.append(num)

    return result

result_list = combine_odd_even(list1, list2)
print(f"result list: {result_list}")
```

5. Display numbers from a list using a loop

Write a Python program to display only those numbers from a list that satisfy the following conditions

The number must be divisible by five
If the number is greater than 150, then skip it and move to the following number
If the number is greater than 500, then stop the loop
Given:

numbers = [12, 75, 150, 180, 145, 525, 50]
Expected output:

75
150
145


```python
numbers = [12, 75, 150, 180, 145, 525, 50]

for num in numbers:
        if num > 500:
            break
        if num > 150:
            continue
        if num % 5 == 0:
            print(num)
```

```python
def display_numbers(numbers):
    for num in numbers:
        if num > 500:
            break
        if num > 150:
            continue
        if num % 5 == 0:
            print(num)

numbers = [12, 75, 150, 180, 145, 525, 50]
display_numbers(numbers)
```

multiplication table of: 1

7. Create a string made of the first, middle, and last character:
Write a program to create a new string made of an input string’s first, middle, and last character.

Given:

str1 = "James"
Expected Output:

Jms

```python
string = str(input("Enter a string: "))

first = string[0]
middle = string[len(string) // 2]
last = string[-1]

result = first + middle + last
print(result)
```

```python
def first_middle_last(text):
    first_char = text[0]
    middle_index = len(text) // 2
    middle_char = text[middle_index]
    last_char = text[-1]
    return first_char + middle_char + last_char

str1 = str(input("Enter a string: "))
result = first_middle_last(str1)
print(result)
```

8. Remove duplicates from a list
Write a function that takes a list with duplicate elements and returns a new list with only unique elements.

Given: list_with_duplicates = [1, 2, 2, 3, 1, 4, 5, 4]

Expected Output:

[1, 2, 3, 4, 5]

```python
list_with_duplicates = [1, 2, 2, 3, 1, 4, 5, 4]
unique_list = list(set(list_with_duplicates))
print(unique_list)
```

```python
def remove_duplicates(numbers):
    unique_numbers = []

    for num in numbers:
        if num not in unique_numbers:
            unique_numbers.append(num)

    return unique_numbers


list_with_duplicates = [1, 2, 2, 3, 1, 4, 5, 4]
result = remove_duplicates(list_with_duplicates)
print(result)
```

Given:

sample_dict = {

    "salary": 8000,
    "city": "New york"
}

# Keys to remove
keys = ["name", "salary"]

```python
sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"
}

keys_to_remove = ["name", "salary"]

for key in keys_to_remove:
    sample_dict.pop(key, None)

print(sample_dict)
```

```python
def remove_keys(dictionary, keys_to_remove):
    for key in keys_to_remove:
        dictionary.pop(key, None)
    return dictionary


sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"
}

keys = ["name", "salary"]

result = remove_keys(sample_dict, keys)
print(result)
```

10. Rename a key in a dictionary

Write a program to rename a key city to a location in the following dictionary.

Given:

sample_dict = {
  "name": "Kelly",
  "age":25,
  "salary": 8000,
  "city": "New york"
}
Expected output:

{'name': 'Kelly', 'age': 25, 'salary': 8000, 'location': 'New york'}

```python
sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"
}

sample_dict["location"] = sample_dict.pop("city")

print(sample_dict)
```

```python
def rename_key(dictionary, old_key, new_key):
    if old_key in dictionary:
        dictionary[new_key] = dictionary.pop(old_key)
    return dictionary


sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"
}

result = rename_key(sample_dict, "city", "location")
print(result)
```

Check if a given year is a leap year
A leap year is a year in the Gregorian calendar that contains an extra day, making it 366 days long instead of the usual 365. This extra day, February 29th, is added to keep the calendar synchronized with the Earth’s revolution around the Sun.

Rules for leap years: a year is a leap year if it’s divisible by 4, unless it’s also divisible by 100 but not by 400.

Write a code find if a given year is a leap year.

Given:

year1 = 2020
# Output True

year2 = 2025
# Output False

```python
def leap_year_checker(year):
  if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    return True
  else:
    return False

year_1 = leap_year_checker(2020)
year_2 = leap_year_checker(2025)

print(year_1)
print(year_2)
```

Write a program to count occurrences of all characters within a string
Given:

str1 = "Apple"
Expected Outcome:

{'A': 1, 'p': 2, 'l': 1, 'e': 1}

```python
def string_counter(text):
    count = {}
    for char in text:
        if char in count:
            count[char] += 1
        else:
            count[char] = 1
    return count

string_text = string_counter("Apple")
print(string_text)
```

Check if a value exists in a dictionary

While we know how to check for a key’s presence in a dictionary, it’s sometimes necessary to determine if a specific value exists.

Write a Python program to check if the value 200 is present in the provided dictionary.

Given:

sample_dict = {'a': 100, 'b': 200, 'c': 300}
Expected output:

200 present in a dict

```python
def value_exists(dictionary, value):
    if value in dictionary.values():
        print(f"{value} present in the dict")
    else:
        print(f"{value} is not present in the dict")

sample_dict = {'a': 100, 'b': 200, 'c': 300}

value_exists(sample_dict, 200)


```
